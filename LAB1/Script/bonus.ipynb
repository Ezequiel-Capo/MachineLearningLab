```python
import pandas as pd

# Cargar dataset
df = pd.read_csv("../Datos_preprocesados/modelo1_vent1.csv")

# Filtrar solo las columnas necesarias
df = df[["fen_carrasco_17", "fase_lunar", "dia_lluvioso"]]

largo = len(df)
print(largo)
# Filtrar casos: trueno + luna nueva
subset = df[(df["fen_carrasco_17"] == "si") & (df["fase_lunar"] == "luna_nueva")]

# Contar ocurrencias según dia_lluvioso
res = subset["dia_lluvioso"].value_counts()

print(res)
```
